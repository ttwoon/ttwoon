## Monday: Lists

```python
# declaring a list of numbers
nums = [5, 10, 15.2, 20]
print(nums)
```

```python
# accessing elements within a list
print(nums[1])  # will output the value at index 1 = 10
num = nums[2]   # saves index value 2 into num
print(num)      # prints value assigned to num
```

```python
# declaring a list of mixed data types
num = 4.3
data = [num, "word", True]   # the power of data collection
print(data)
```

```python
# understanding lists within lists
data = [5, "book", [34, "hello"], True]
print(data)
print(data[2])
```

```python
# using double bracket notation to access lists within lists
print(data[2][0])
inner_list = data[2]
print(inner_list[1])
```

```python
# changing values in a list through index
data = [5, 10, 15,20]
print(data)
data[0] = 100
print(data)
```

```python
# understanding how lists are stored
a = [5, 10]
b = a
print(f"a: {a}\t b: {b}")
print(f"Location a[0]: {id(a[0])}\t Location b[0]: {id(b[0])}")
a[0] = 20
print(f"a: {a}\t b: {b}")
```

```python
# using [:] to copy
data = [5, 10, 15, 20]
data_copy = data[ : ]
data[0] = 50
print(f"data: {data}\t data_copy: {data_copy}")

```

# Tuesday: For Loops

```python
# writing your first for loop using range
for num in range(5):
  print(f"Value: {num}")
```

```python
# providing the start, stop amd step for the range function
for num in range(2, 10, 2):
  print(f"Value: {num}")
```

```python
# printing all characters in a name using the "in" keyword
name = "John Smith"
for letter in name:
  print(f"Value: {letter}")
```

```python
# using the continue statement within a for loop
for num in range(5):
  if num == 3:
    continue
  print(num)
```

```python
# breaking out of a loop using the 'break' keyword
for num in range(5):
  if num == 3:
    break
  print(num)
```

```python
# setting a placeholder using the 'pass' keyword
for i in range(5):
  # TODO: add code to print number
  pass
```

## Wednesday: While Loop

```python
# writing your first while loop
health = 10
while health > 0:
  print(health)
  health -= 1

```

```python
# using two or more loops together is called a nested loop
for i in range(2):
  for j in range(3):
    print(i, j)
```

## Thursday: Working with Lists

```python
# checking the number of items within a list
nums = [5, 10, 15]
length = len(nums)
print(length)
```

```python
# accessing specific items of a list with slices
print(nums[1:3])
print(nums[:2])
print(nums[::2])
print(nums[-2::])
```

```python
# adding an item to the back of a list using append
nums = [10,20]
nums.append(5)
print(nums)
```

```python
# adding a value to the begining of a list
words = ["balls", "base"]
words.insert(0, "glove")
print(words)
```

```python
# using pop to remove items and saving to a variable to use later
items = [5, "ball", True]
items.pop()
print(items)
removed_item = items.pop(0)
print(removed_item, "\n", items)
```

```python
# using the remove method with a try and except
sports = ["baseball", "soccer", "football", "hockey"]
try:
  sports.remove("soccer")
except:
  print(f"That item does not exit in the list")
print(sports)
```

```python
# using min, max and sum
nums = [5, 3, 9]
print(min(nums))
print(max(nums))
print(sum(nums))
```

```python
# using sorted on lists for numerical and alphabetical data
nums = [5, 8, 0, 2]
sorted_nums = sorted(nums)
print(nums, sorted_nums)
```

```python
# sorting a list with .sort() in-place
nums = [5, 0, 8, 3]
nums.sort()
print(nums)
```

```python
# using conditional statements on a list
names = ["Jack", "Robert", "Mary"]
if "Mary" in names:
  print("found")
if "Jimmy" not in names:
  print("not found")
```

```python
# using conditionals to see if a list is empty
nums = [ ]
if not nums:
  print("empty")
```

```python
# using a for loop to print all items in a list
sports = ["Baseball", "Hockey", "Football", "Basketball"]
for sport in sports:
  print(sport)
```

```python
# using the while loop to remove a certain value
names = ["Bob", "Jack", "Rob", "Bob", "Robert"]
while "Bob" in names:
  names.remove("Bob")
print(names)
```
