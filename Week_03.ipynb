# Logical Operators

## Logical Operator "and"

```python
# using the keyword 'an' in an 'if statement'
x, y, z = 5, 10, 5
if x < y and x == z:
    print("Both statements are true")
```

Logical Operator "or"


```python
# using the keyword 'or' in an 'if statement'
x, y, z = 5, 10, 5
if x < y or x != z:
    print("One or both statements were true.")
```

Logical Operator "not"


```python
# using the keyword 'not' within an 'if statement'
flag = False
if not flag:
    print("Flag is False")
```

Membership Operators

Membership Operator "in"

```python
# using the keyword 'in' within an 'if statement'
word = "Baseball"
if "b" in word:
    print("{} contain the character b".format(word))

```

Membership Operator "not in"

```python
# using the keyword 'not in' within an 'if statement'
word = "Baseball"
if "x" not in word:
    print("{} does not contain the character x".format(word))
```

# Friday: Creating a Calculator

```python
# step 1: ask user for calculation to be performed
operation = input("What calculation would you like to perform? (add, subtract, multiply, divide): ").lower()

# step 2: ask for numbers, Alert Order Matters
if operation == "subtract" or operation == "divide":
    print("You chose {}".format(operation))
    print("Order of numbers matters for this operation.")
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
print("First number: {}".format(num1))
print("Second number: {}".format(num2))

# step 3: Setup Try / Except for Mathematical Operations
try:
    num1, num2 = float(num1), float(num2)
    if operation == "add":
        result = num1 + num2
        print("{} + {} = {}".format(num1, num2, result))
    elif operation == "subtract":
        result = num1 - num2
        print("{} - {} = {}".format(num1, num2, result))
    elif operation == "multiply":
        result = num1 * num2
        print("{} * {} = {}".format(num1, num2, result))
    elif operation == "divide":
        result = num1 / num2
        print("{} / {} = {}".format(num1, num2, result))
    else:
        print("Sorry, but '{}' is not an option.".format(operation))
except:
    print("Error: Improper numbers used. Please enter valid numbers.")
```
